```python
import pandas
import json
```

```python
re_path = "results/scifact/kimi-k2-0711/R__bge-large-en-v1_5_k10-RR__bge-reranker-large_n5-EX__bge-large-en-v1_5/WBTQ_RW-0_RR-1_EX-0_IF-0_OF-0_repeat_command_0.jsonl"
```

```python
def jsonl_results_loader(save_path,num_records=500):
    """加载 JSONL 格式的结果文件"""
    results = []
    with open(save_path, "r", encoding="utf-8") as f:
        for line in f:
            record = json.loads(line.strip())
            results.append(record)
    return results[0:num_records]
```

```python
data_jsonl = jsonl_results_loader(re_path)
```

```python
data={
        "queries": [item['cleaned_query'] for item in data_jsonl],
        "answers": [item['answer'] for item in data_jsonl],
        "contexts": [item['contexts'] for item in data_jsonl]
    }
```
