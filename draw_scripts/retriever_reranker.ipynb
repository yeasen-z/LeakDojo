```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

metric_col = "rougeL_05_chunk"

retriever_order = ["BM25:1", "all-MiniLM-L6-v2:2", "all-mpnet-base-v2:3", "bge-large-en-v1.5:4"]
reranker_order = ["None:1", "ms-marco-TinyBERT-L-2:2", "ms-marco-MiniLM-L-6-v2:3", "bge-reranker-large:4"]

models_to_plot = ["gemma-3-12b-it", "Qwen2.5-14B-Instruct"]
linestyles = ['-', '--']
colors = ['tab:blue', 'tab:orange', 'tab:green', 'tab:red']

x = np.arange(len(retriever_order))

fig, axes = plt.subplots(1, 2, figsize=(14, 6), sharey=False)  # 两个子图，纵轴不共享

for model_idx, model_name in enumerate(models_to_plot):
    ax = axes[model_idx]
    df_model = df[df["model"] == model_name]
    pivot = df_model.pivot_table(index="retriever", columns="reranker", values=metric_col)
    pivot = pivot.reindex(index=retriever_order, columns=reranker_order)

    for reranker_idx, reranker in enumerate(pivot.columns):
        ax.plot(x, pivot[reranker], marker='o', linestyle=linestyles[model_idx],
                color=colors[reranker_idx], label=reranker)

    ax.set_xticks(x)
    ax.set_xticklabels(retriever_order, rotation=20)
    ax.set_xlabel("Retriever")
    ax.set_ylabel(metric_col)
    ax.set_title(f"{model_name}")
    ax.grid(True)
    ax.legend(title="Reranker", loc='lower right')

plt.suptitle(f"Comparison of {metric_col} Across Models", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```
