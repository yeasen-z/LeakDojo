```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from datetime import datetime
import matplotlib.colors as mcolors

# ==========================================
# 1. ACL 全局风格配置
# ==========================================
plt.rcParams.update({
    'font.family': 'sans-serif',
    'font.sans-serif': ['Nimbus Sans'],
    'font.size': 7,
    'axes.labelsize': 10,
    'axes.titlesize': 11,
    'legend.fontsize': 8,
    'xtick.labelsize': 9,
    'ytick.labelsize': 9,
    'figure.dpi': 200,
    'savefig.bbox': 'tight',
    'axes.unicode_minus': False
})

# ACL 蓝色系
C_DEEP = "black"     # 深蓝色(线)
C_DEEP_dot = "#1f77b4"      # 橙色(点)
C_LIGHT = "black"    # 浅蓝色(置信区间) 
C_GRID = "#e0e0e0"     # 浅灰色(网格)

# ==========================================
# 2. 数据准备
# ==========================================
models = [
    "Qwen-2.5-7B", "Qwen-2.5-14B", "Qwen-2.5-32B", "Qwen-2.5-72B",
    "o4-mini", "DeepSeek-V3", "Kimi-K2", "Gemma-3-27B",
    "Gemini-2.5-flash", "Doubao-1.6-flash",
    "Qwen-3-235B", "Gemini-3-flash", "GPT-5.1"
]
ifeval = np.array([71.9, 89.83, 78.93, 83.73, 91.5, 76.16, 82.81, 85.25, 90.38, 76.89, 85.21, 92.53, 86.32])
risk = np.array([64.5, 70.52, 54.4, 54.46, 67.75, 46.85, 64.68, 80.2, 80.3, 68.8, 59.92, 76.67, 72.42])

# ==========================================
# 2.1 Model Release Dates (for color coding freshness)
# ==========================================
# Release dates for each model (YYYY-MM-DD format)
release_dates = {
    "Qwen-2.5-7B": "2024-09-19",
    "Qwen-2.5-14B": "2024-09-19",
    "Qwen-2.5-32B": "2024-09-19",
    "Qwen-2.5-72B": "2024-09-19",
    "o4-mini": "2025-04-16",
    "DeepSeek-V3": "2024-12-26",
    "Kimi-K2": "2025-07-01",
    "Gemma-3-27B": "2025-03-12",
    "Gemini-2.5-flash": "2025-03-25",
    "Doubao-1.6-flash": "2025-06-15",
    "Qwen-3-235B": "2025-04-29",
    "Gemini-3-flash": "2025-11-15",
    "GPT-5.1": "2025-10-28"
}

# Calculate days since release (relative to current date: Jan 4, 2026)
current_date = datetime(2026, 1, 4)
days_since_release = []
for model in models:
    release = datetime.strptime(release_dates[model], "%Y-%m-%d")
    days = (current_date - release).days
    days_since_release.append(days)

days_since_release = np.array(days_since_release)

# Normalize to 0-1 range for colormap
# Newer models (fewer days) should be darker, older models should be lighter
norm_days = (days_since_release - days_since_release.min()) / (days_since_release.max() - days_since_release.min())

# Use single-color gradient: Blues gives darker blue for larger values (older = lighter, newer = darker)
# We'll invert the mapping so older (more days) = left = lighter, newer (fewer days) = right = darker
cmap = plt.cm.Blues
colors = [cmap(d) for d in norm_days]

# 2:1 宽高比
fig, ax = plt.subplots(figsize=(3, 2))

# ==========================================
# 3. 绘制回归趋势与置信区间
# ==========================================
# 先画阴影和趋势线
sns.regplot(
    x=ifeval, y=risk, ax=ax,
    scatter=False, 
    color=C_LIGHT, 
    line_kws={'color': C_DEEP, 'linewidth': 1, 'alpha': 0.8, 'zorder': 2, 'linestyle': '--'},
    ci=95,
    scatter_kws={'alpha': 0.5}
)

# 后画散点,确保点在最上方 - color by freshness
# Invert the color mapping: older (more days) should be lighter, newer (fewer days) should be darker
# Invert days so that newer (fewer days) maps to higher values = darker color
min_days, max_days = days_since_release.min(), days_since_release.max()
inverted_days = max_days - days_since_release + min_days
scatter = ax.scatter(
    ifeval, risk, 
    c=inverted_days,       # Higher value = newer = darker blue
    cmap='Blues',          # Left (low) = older/lighter, Right (high) = newer/darker
    s=30, 
    edgecolors='black', 
    linewidths=0.4, 
    alpha=0.95, 
    zorder=5,
    vmin=inverted_days.min(),  # Explicitly set color range
    vmax=inverted_days.max()
)

# ==========================================
# 4. 智能标注优化
# ==========================================
# 为每个模型定义精确的偏移量以避免重叠
label_offsets = {
#     "Qwen-2.5-7B": (6, 0, 'left'),
#     "Qwen-2.5-14B": (6, 8, 'left'),
    "Qwen-2.5-32B": (0, 6, 'left'),
#     "Qwen-2.5-72B": (6, -8, 'left'),
#     "o4-mini": (6, 12, 'left'),
#     "DeepSeek-V3": (6, 0, 'left'),
#     "Kimi-K2": (6, 8, 'left'),
    "Gemma-3-27B": (-6, 0, 'right'),
#     "Gemini-2.5-flash": (6, 0, 'left'),
#     "Doubao-1.6-flash": (6, -8, 'left'),
#     "Qwen-3-8B": (6, 12, 'left'),
#     "Qwen-3-235B": (6, -8, 'left'),
    "Gemini-3-flash": (-6, 0, 'right'),
    "Gemini-2.5-flash": (20, 6, 'right'),
#     "GPT-5.1": (6, -10, 'left')
}

# label_offsets = {}

for i, name in enumerate(models):
    # 进一步简化模型名
    short_name = name
    
    # short_name = name
    # 获取预定义的偏移量
    dx, dy, ha = label_offsets.get(name, (6, 0, 'left'))

    ax.annotate(
        short_name,
        (ifeval[i], risk[i]),
        xytext=(dx, dy),
        textcoords='offset points',
        color="#666666",
        fontsize=5,
        ha=ha,
        va='center',
        # 给文字加一个极淡的白色光晕背景,确保经过线时清晰
        bbox=dict(boxstyle='round,pad=0.2', fc='white', alpha=0.7, ec='none')
    )

# ==========================================
# 5. 轴美化与修饰
# ==========================================
ax.set_xlabel('IFEval Strict Score (%)', labelpad=2, fontsize=6)
ax.set_ylabel('Probability of Leakage (%)', labelpad=2, fontsize=6)
ax.tick_params(axis='both', labelsize=6)
# ax.set_title('Instruction Following Ability vs. Leakage Risk', pad=15, fontweight='bold', fontsize=11)

# 隐藏顶部和右侧边框
sns.despine()

# 设置细腻的网格
ax.grid(True, which='major', linestyle='--', color=C_GRID, alpha=0.6, zorder=0)

# 限制坐标范围,增加呼吸感
ax.set_xlim(70, 96)
ax.set_ylim(40, 92)

# Call tight_layout BEFORE creating inset axes to avoid compatibility issues
plt.tight_layout()

# ==========================================
# 5.1 添加颜色条表示模型新鲜度 - 放在右上角
# ==========================================
# Create a small inset axes for the colorbar in the upper right
# Use fig.add_axes() instead of inset_axes to avoid tight_layout issues
# Position: [left, bottom, width, height] in figure coordinates
cax = fig.add_axes([0.55, 0.87, 0.25, 0.025])
cbar = plt.colorbar(scatter, cax=cax, orientation='horizontal')
cbar.ax.tick_params(labelsize=6)
# Left = older (low inverted_days, lighter blue), Right = newer (high inverted_days, darker blue)
# inverted_days: min = oldest, max = newest
inv_min, inv_max = inverted_days.min(), inverted_days.max()
cbar.set_ticks([])
cax.xaxis.set_ticks_position('bottom')
cax.xaxis.set_label_position('top')
# Add the newest date at the right of the colorbar
# inverted_days.max() corresponds to the newest model
newest_idx = int(inverted_days.argmax())
# Use models list to get the model name, then lookup in release_dates
newest_model = models[newest_idx]
newest_date = release_dates[newest_model]
cax.text(1.05, 0.5, "2025.11", transform=cax.transAxes, fontsize=5.5, 
         va='center', ha='left', color='black')

# 添加相关系数标注
from scipy.stats import pearsonr
corr, p_value = pearsonr(ifeval, risk)
ax.text(0.05, 0.95, f'Pearson r = {corr:.3f}\np < 0.001' if p_value < 0.001 else f'Pearson r = {corr:.3f}\np = {p_value:.3f}',
        transform=ax.transAxes, fontsize=6, verticalalignment='top')
from matplotlib.offsetbox import AnchoredOffsetbox, TextArea, HPacker, VPacker

title_main = "Stronger instruction-following ability can imply\nhigher leakage risk of RAG systems."

# clear the normal title (we'll draw our own)
ax.set_title("")


# ==========================================
# 6. 添加方向箭头
# ==========================================
# 水平箭头 - 指示 "Stronger instruction-following"
arrow_x_start = 82.5
arrow_x_end = 94
arrow_y = 44

ax.annotate(
    '',
    xy=(arrow_x_end, arrow_y),
    xytext=(arrow_x_start, arrow_y),
    arrowprops=dict(
        arrowstyle='->',
        color='#1f77b4',
        lw=1.5,
        alpha=0.8
    ),
    zorder=3
)

# 水平箭头文字标注
ax.text(
    (arrow_x_start + arrow_x_end) / 2,
    arrow_y + 8,
    'Stronger\ninstruction-following',
    ha='center',
    va='top',
    fontsize=6,
    color='#1f77b4',
    weight='bold'
)

# 垂直箭头 - 指示 "Higher risk"
arrow_x = 95
arrow_y_start = 48
arrow_y_end = 68

ax.annotate(
    '',
    xy=(arrow_x, arrow_y_end),
    xytext=(arrow_x, arrow_y_start),
    arrowprops=dict(
        arrowstyle='->',
        color='#d62628',
        lw=1.5,
        alpha=0.8
    ),
    zorder=3
)

# 垂直箭头文字标注
ax.text(
    arrow_x - 1.9,
    (arrow_y_start + arrow_y_end) / 2 - 1,
    'Higher\nrisk',
    ha='center',
    va='center',
    fontsize=6,
    color='#d62628',
    weight='bold'
)

plt.savefig('ifeval_vs_leakage_trend_freshness.pdf', dpi=300, facecolor='white')
# plt.show()
```
