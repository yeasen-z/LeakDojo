```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# 确保分类顺序正确
scale_order = ['1.5B', '3-4B', '7-8B', '14B', '32B']
df_copy['model_scale'] = pd.Categorical(df_copy['model_scale'], categories=scale_order, ordered=True)

# 1️⃣ 聚合：计算每个系列在每个模型规模下的均值和标准差
df_summary = (
    df_copy.groupby(['model_series', 'model_scale'])
    .agg(mean=('rougeL_05_chunk', 'mean'),
         std=('rougeL_05_chunk', 'std'))
    .reset_index()
)

# 2️⃣ 设置颜色调色板
series_list = df_summary['model_series'].unique()
palette = sns.color_palette("Set2", n_colors=len(series_list))
color_map = dict(zip(series_list, palette))

# 3️⃣ 绘制趋势 + 波动带
plt.figure(figsize=(10, 6))
sns.set(style="whitegrid", font_scale=1.1)

for s in series_list:
    d = df_summary[df_summary['model_series'] == s].sort_values('model_scale')
    
    # 画平均趋势线
    plt.plot(d['model_scale'], d['mean'], label=s, color=color_map[s], linewidth=2.5)
    
    # 画波动带（±1 std）
    plt.fill_between(
        d['model_scale'],
        d['mean'] - d['std'],
        d['mean'] + d['std'],
        color=color_map[s],
        alpha=0.25
    )

# 4️⃣ 美化
plt.title("RougeL_05_chunk Trend (Mean ± Std)", fontsize=14, pad=15)
plt.xlabel("Model Scale", fontsize=12)
plt.ylabel("RougeL_05_chunk", fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.legend(title="Model Series", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
